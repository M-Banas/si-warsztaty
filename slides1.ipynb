## Stan początkowy i zasady

:::: {.columns}

::: {.column width="48%"}
| Parametr gry | Wartość początkowa |
|---|---|
| Monety gracza (`coins`) | **15 monet** |
| Zapas kupca (`potions`) | **2 mikstury** |
| Cena mikstury (`potion_price`) | **10 monet** / szt. |

<br>

**Dozwolone akcje (`ALLOWED_ACTIONS`):**

* `"talk"` – zwykła rozmowa, pytanie o ofertę, odmowa.
* `"buy_potion"` – intencja zakupu dokładnie **1 mikstury**.
:::

::: {.column width="52%"}
**Format odpowiedzi modelu (JSON):**

Wymuszany przez `MerchantResponse` (`Pydantic` + `ConfigDict(extra="forbid")`, `json_schema` w OpenAI Responses API):

```json
{
  "reply": "Tekst odpowiedzi kupca w świecie gry...",
  "action": "talk"
}
```

Niepoprawny JSON, nieznana akcja spoza `ALLOWED_ACTIONS` lub przekroczenie czasu oczekiwania (`timeout = 20 s`) **nigdy nie zmieniają stanu gry**. Komunikat o transakcji powstaje na podstawie decyzji programu, a nie samej wypowiedzi `reply`.
:::
